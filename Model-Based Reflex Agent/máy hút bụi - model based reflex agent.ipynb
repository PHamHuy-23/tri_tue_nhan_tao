# Máy Hút Bụi (Tác Nhân Phản Xạ Dựa Trên Mô Hình)

## Nhập Thư Viện và Thiết Lập Môi Trường

```python
import random
import time

# =========================================
# MÔI TRƯỜNG
# 1 = BẨN
# 0 = SẠCH
# =========================================

room = [
    [1, 0, 1],
    [1, 1, 0],
    [0, 1, 1]
]

ROWS = len(room)
COLS = len(room[0])

# =========================================
# VỊ TRÍ BẮT ĐẦU CỦA TÁC NHÂN
# =========================================

x = 0
y = 0

# =========================================
# QUY TẮC
# =========================================

rules = {
    0 : "DI CHUYỂN",
    1 : "HÚT BỤI"
}

# =========================================
# MÔ HÌNH / BỘ NHỚ
# =========================================

visited = []

cleaned = []
```

## Hàm Hiển Thị Phòng

```python
# =========================================
# HIỂN THỊ PHÒNG
# =========================================

def hien_thi_phong():
    print("\nTRẠNG THÁI PHÒNG HIỆN TẠI\n")
    for i in range(ROWS):
        for j in range(COLS):
            # vị trí tác nhân
            if i == x and j == y:
                print("A", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
```

## Hàm Quét Vị Trí Hiện Tại

```python
# =========================================
# QUÉT VỊ TRÍ HIỆN TẠI
# =========================================

def quet_vi_tri_hien_tai():
    print("\nQUÉT VỊ TRÍ HIỆN TẠI")
    print("Vị trí Tác nhân :", (x, y))
    print("Trạng thái Ô :", room[x][y])
```

## Hàm Cập Nhật Mô Hình

```python
# =========================================
# CẬP NHẬT MÔ HÌNH
# =========================================

def cap_nhat_mo_hinh():
    global visited, cleaned
    # lưu vị trí đã ghé thăm
    if (x, y) not in visited:
        visited.append((x, y))
    # lưu vị trí đã làm sạch
    if room[x][y] == 0 and (x, y) not in cleaned:
        cleaned.append((x, y))
    print("\nMÔ HÌNH ĐÃ ĐƯỢC CẬP NHẬT")
    print("Đã ghé thăm :", visited)
    print("Đã làm sạch :", cleaned)
```

## Hàm Hành Động Hút Bụi

```python
# =========================================
# HÀNH ĐỘNG HÚT BỤI
# =========================================

def hut_bui():
    global room
    print("\nHÀNH ĐỘNG : HÚT BỤI")
    room[x][y] = 0
    print("BỤI ĐÃ ĐƯỢC LOẠI BỎ")
```

## Hàm Hành Động Di Chuyển

```python
# =========================================
# HÀNH ĐỘNG DI CHUYỂN
# =========================================

def di_chuyen():
    global x, y
    possible_moves = []

    # LÊN
    if x > 0:
        possible_moves.append(("LÊN", x - 1, y))

    # XUỐNG
    if x < ROWS - 1:
        possible_moves.append(("XUỐNG", x + 1, y))

    # TRÁI
    if y > 0:
        possible_moves.append(("TRÁI", x, y - 1))

    # PHẢI
    if y < COLS - 1:
        possible_moves.append(("PHẢI", x, y + 1))

    print("\nCÁC BƯỚC DI CHUYỂN CÓ THỂ")
    print(possible_moves)

    # =====================================
    # ƯU TIÊN:
    # DI CHUYỂN ĐẾN Ô CHƯA GHÉ THĂM
    # =====================================

    unvisited = []

    for direction, nx, ny in possible_moves:
        if (nx, ny) not in visited:
            unvisited.append((direction, nx, ny))

    # nếu có các ô chưa ghé thăm
    if len(unvisited) > 0:
        direction, nx, ny = random.choice(unvisited)
        print("\nCHỌN Ô CHƯA GHÉ THĂM")

    else:
        direction, nx, ny = random.choice(possible_moves)
        print("\nTẤT CẢ CÁC Ô ĐÃ GHÉ THĂM")
        print("DI CHUYỂN NGẪU NHIÊN")

    print("ĐÃ CHỌN :", direction)

    # cập nhật vị trí
    x = nx
    y = ny
```

## Hàm Kiểm Tra Phòng Đã Sạch Chưa

```python
# =========================================
# KIỂM TRA PHÒNG ĐÃ SẠCH CHƯA
# =========================================

def phong_da_sach_chua():
    for row in room:
        for cell in row:
            if cell == 1:
                return False
    return True
```

## Hàm Tác Nhân Phản Xạ Dựa Trên Mô Hình

```python
# =========================================
# TÁC NHÂN PHẢN XẠ DỰA TRÊN MÔ HÌNH
# =========================================

def tac_nhan_phan_xa_dua_tren_mo_hinh():
    # bước 1 : hiển thị phòng
    hien_thi_phong()

    # bước 2 : quét môi trường
    quet_vi_tri_hien_tai()

    # bước 3 : cập nhật mô hình nội bộ
    cap_nhat_mo_hinh()

    # bước 4 : lấy trạng thái hiện tại
    state = room[x][y]
    action = rules[state]

    print("\nQUY TẮC ĐƯỢC CHỌN :", action)

    # =====================================
    # HÚT BỤI
    # =====================================

    if action == "HÚT BỤI":
        hut_bui()
        cap_nhat_mo_hinh()

    # =====================================
    # DI CHUYỂN
    # =====================================

    di_chuyen()
```

## Thực Thi Chương Trình Chính

```python
# =========================================
# CHƯƠNG TRÌNH CHÍNH
# =========================================

step = 1

while True:
    print("\n=================================")
    print("BƯỚC :", step)
    print("=================================")

    # điều kiện dừng
    if phong_da_sach_chua():
        print("\nTẤT CẢ CÁC PHÒNG ĐỀU SẠCH")
        hien_thi_phong()
        break

    # chạy tác nhân
    tac_nhan_phan_xa_dua_tren_mo_hinh()

    step += 1

    time.sleep(1)
```
