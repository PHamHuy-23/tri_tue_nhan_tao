# 🛡️ Forward Checking (Kiểm tra phía trước) - Tô màu bản đồ Úc

Forward Checking là cải tiến quan trọng của Backtracking Search.
Mỗi khi một biến được gán giá trị, thuật toán sẽ kiểm tra các biến lân cận chưa gán và **loại bỏ** các giá trị không còn hợp lệ khỏi miền giá trị (domain) của chúng.
Nếu domain của bất kỳ biến chưa gán nào bị rỗng, thuật toán quay lui ngay lập tức mà không cần tiến sâu hơn, giúp tiết kiệm đáng kể số bước duyệt.
