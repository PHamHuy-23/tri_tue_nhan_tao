# ⚖️ AC-3 (Arc Consistency) - Nhất quán cung - Tô màu bản đồ Úc

AC-3 giúp thu hẹp đáng kể miền giá trị của các biến, thậm chí phát hiện sớm trường hợp vô nghiệm.
