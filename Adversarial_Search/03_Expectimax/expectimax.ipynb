# 🎲 Expectimax Algorithm - Tic-Tac-Toe

Expectimax được sử dụng khi môi trường hoặc đối thủ có yếu tố ngẫu nhiên (không chơi tối ưu tuyệt đối mà chọn nước đi ngẫu nhiên với xác suất).
Thay vì lấy giá trị Min như Minimax, Expectimax tại nút đối thủ sẽ tính **Giá trị kỳ vọng (Expected Value)** của các nước đi.

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
sys.path.append('../../Search_Algorithm')
from adversarial_solvers import expectimax

board = [0] * 9
```

## 🔍 2. Chạy thuật toán Expectimax

```python
stats = {"nodes": 0}
score, move = expectimax(board, 0, True, stats)

print("=== KẾT QUẢ EXPECTIMAX ===")
print(f"Giá trị kỳ vọng của bàn cờ ban đầu: {score:.4f}")
print(f"Nước đi tốt nhất cho X: Ô số {move}")
print(f"Tổng số nút đã duyệt: {stats['nodes']}")
```
