# ✂️ Alpha-Beta Pruning (Cắt tỉa Alpha-Beta) - Tic-Tac-Toe

Alpha-Beta Pruning là phương pháp tối ưu hóa cực kỳ quan trọng cho Minimax.
Bằng cách duy trì hai giá trị $\alpha$ (giá trị tốt nhất MAX chắc chắn đạt được) và $\beta$ (giá trị tốt nhất MIN chắc chắn giữ được), thuật toán cắt bỏ các nhánh con không làm thay đổi quyết định cuối cùng, giúp giảm đáng kể số nút cần duyệt.

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
sys.path.append('../../Search_Algorithm')
from adversarial_solvers import alpha_beta

board = [0] * 9
```

## 🔍 2. Chạy thuật toán Alpha-Beta và so sánh số nút duyệt với Minimax

```python
stats = {"nodes": 0, "prunes": 0}
score, move = alpha_beta(board, 0, -float('inf'), float('inf'), True, stats)

print("=== KẾT QUẢ ALPHA-BETA PRUNING ===")
print(f"Điểm lượng giá tối ưu: {score}")
print(f"Nước đi tốt nhất cho X: Ô số {move}")
print(f"Tổng số nút đã duyệt: {stats['nodes']} (So với Minimax: 549,946 nút)")
print(f"Số lần thực hiện cắt tỉa nhánh: {stats['prunes']}")
```
