# ⚔️ Minimax Algorithm - Tic-Tac-Toe

Minimax là thuật toán tìm kiếm đối kháng kinh điển cho trò chơi hai người chơi đối kháng toàn diện (được ăn cả ngã về không).
Max (người chơi X) cố gắng tối đa hóa điểm lượng giá, trong khi Min (người chơi O) cố gắng tối thiểu hóa điểm lượng giá.

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
sys.path.append('../../Search_Algorithm')
from adversarial_solvers import minimax

# Khởi tạo bàn cờ Tic-Tac-Toe trống
board = [0] * 9
```

## 🔍 2. Chạy thử thuật toán Minimax

```python
stats = {"nodes": 0}
score, move = minimax(board, 0, True, stats)

print("=== KẾT QUẢ MINIMAX ===")
print(f"Điểm lượng giá tối ưu của trạng thái ban đầu: {score}")
print(f"Nước đi tốt nhất cho X (MAX): Ô số {move}")
print(f"Tổng số nút đã duyệt (Search Space): {stats['nodes']}")
```
