# 8 Puzzle Simple Reflex Agent

## PEAS – 8 Puzzle Simple Reflex Agent

### Performance Measure
- Đưa các ô số về đúng vị trí
- Di chuyển hợp lệ
- Không đi ra ngoài ma trận

### Environment
- Ma trận 3x3
- Gồm các số 1 → 8
- Một ô trống = 0

**Ví dụ:**

```
1 2 3
4 0 5
6 7 8
```

### Actuators
- Move Up
- Move Down
- Move Left
- Move Right

### Sensors
- Kiểm tra vị trí ô trống
- Kiểm tra trạng thái hiện tại
- Kiểm tra biên ma trận

## 2. RULE TABLE

| Điều kiện           | Hành động          |
| :----------------- | :----------------- |
| Có thể đi lên      | Random chọn UP     |
| Có thể đi xuống    | Random chọn DOWN   |
| Có thể đi trái     | Random chọn LEFT   |
| Có thể đi phải     | Random chọn RIGHT  |
| Chạm biên          | Loại hướng đó khỏi danh sách |

**Simple Reflex Agent ở đây:**

- chỉ nhìn trạng thái hiện tại,
- không nhớ bước cũ,
- không tìm đường tối ưu.

## 3. CODE – SIMPLE REFLEX AGENT 8 PUZZLE

````python
import random

# ================= PUZZLE =================

puzzle = [
    [1,2,3],
    [4,0,5],
    [6,7,8]
]

# 0 = empty space

ROWS = 3
COLS = 3

# ==========================================

def show_puzzle():

    print("\nCURRENT PUZZLE\n")

    for row in puzzle:

        for cell in row:

            print(cell, end=" ")

        print()

# ==========================================

def find_empty():

    for i in range(ROWS):

        for j in range(COLS):

            if puzzle[i][j] == 0:

                return i, j

# ==========================================

def move():

    x, y = find_empty()

    print("\nEMPTY POSITION :", (x,y))

    directions = ["w","a","s","d"]

    # remove impossible moves

    if x == 0:
        if "w" in directions: directions.remove("w")

    if x == 2:
        if "s" in directions: directions.remove("s")

    if y == 0:
        if "a" in directions: directions.remove("a")

    if y == 2:
        if "d" in directions: directions.remove("d")

    print("AVAILABLE MOVES :", directions)

    if not directions:
        print("No available moves. Skipping this step.")
        return

    direction = random.choice(directions)

    print("SELECTED MOVE :", direction)

    # ================= UP =================

    if direction == "w":

        puzzle[x][y] = puzzle[x-1][y]

        puzzle[x-1][y] = 0

        print("MOVE UP")

    # ================= DOWN =================

    elif direction == "s":

        puzzle[x][y] = puzzle[x+1][y]

        puzzle[x+1][y] = 0

        print("MOVE DOWN")

    # ================= LEFT =================

    elif direction == "a":

        puzzle[x][y] = puzzle[x][y-1]

        puzzle[x][y-1] = 0

        print("MOVE LEFT")

    # ================= RIGHT =================

    elif direction == "d":

        puzzle[x][y] = puzzle[x][y+1]

        puzzle[x][y+1] = 0

        print("MOVE RIGHT")

# ==========================================

def reflex_agent():

    print("\nSCAN CURRENT STATE")

    show_puzzle()

    print("\nRULE : RANDOM VALID MOVE")

    move()

# ==========================================

for step in range(10):

    print("\n======================")
    print("STEP :", step + 1)
    print("======================")

    reflex_agent()

    show_puzzle()
````

### 3.1. Puzzle Initialization

This cell defines the initial state of the 8-puzzle and its dimensions.

````python
import random

puzzle = [
    [1,2,3],
    [4,0,5],
    [6,7,8]
]

ROWS = 3
COLS = 3

print("Initial Puzzle State:")
for row in puzzle:
    print(row)
````

### 3.2. `show_puzzle` Function

This function is responsible for printing the current state of the puzzle in a readable 3x3 grid format.

````python
def show_puzzle():
    print("\nCURRENT PUZZLE\n")
    for row in puzzle:
        for cell in row:
            print(cell, end=" ")
        print()
````

### 3.3. `find_empty` Function

This function locates the position of the empty tile (represented by `0`) within the puzzle and returns its row and column indices.

````python
def find_empty():
    for i in range(ROWS):
        for j in range(COLS):
            if puzzle[i][j] == 0:
                return i, j
````

### 3.4. `move` Function

This function determines the possible moves for the empty tile, randomly selects one, and then updates the puzzle state accordingly. It considers border constraints to remove impossible moves.

````python
def move():
    x, y = find_empty()
    print("\nEMPTY POSITION :", (x,y))
    directions = ["w","a","s","d"]

    # remove impossible moves
    if x == 0:
        if "w" in directions: directions.remove("w")
    if x == 2:
        if "s" in directions: directions.remove("s")
    if y == 0:
        if "a" in directions: directions.remove("a")
    if y == 2:
        if "d" in directions: directions.remove("d")

    print("AVAILABLE MOVES :", directions)

    if not directions:
        print("No available moves. Skipping this step.")
        return

    direction = random.choice(directions)
    print("SELECTED MOVE :", direction)

    if direction == "w": # UP
        puzzle[x][y], puzzle[x-1][y] = puzzle[x-1][y], puzzle[x][y]
        print("MOVE UP")
    elif direction == "s": # DOWN
        puzzle[x][y], puzzle[x+1][y] = puzzle[x+1][y], puzzle[x][y]
        print("MOVE DOWN")
    elif direction == "a": # LEFT
        puzzle[x][y], puzzle[x][y-1] = puzzle[x][y-1], puzzle[x][y]
        print("MOVE LEFT")
    elif direction == "d": # RIGHT
        puzzle[x][y], puzzle[x][y+1] = puzzle[x][y+1], puzzle[x][y]
        print("MOVE RIGHT")
````

### 3.5. `reflex_agent` Function

This function encapsulates a single step of the simple reflex agent, which involves scanning the current puzzle state, showing it, applying the random valid move rule, and then executing the move.

````python
def reflex_agent():
    print("\nSCAN CURRENT STATE")
    show_puzzle()
    print("\nRULE : RANDOM VALID MOVE")
    move()
````

### 3.6. Simulation Loop

This loop runs the `reflex_agent` for a specified number of steps (10 in this case), displaying the puzzle state before and after each step.

````python
print("\n====== Starting 8-Puzzle Simulation ====== ")
for step in range(10):
    print("\n======================")
    print("STEP :", step + 1)
    print("======================")
    reflex_agent()
    show_puzzle()
print("\n====== Simulation Finished ====== ")
````
