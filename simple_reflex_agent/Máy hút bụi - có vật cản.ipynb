# SIMPLE REFLEX AGENT - MÁY HÚT BỤI

## PEAS

- P -> ĐỘ ĐO HIỆU SUẤT
    - Hút sạch bụi
    - Di chuyển hợp lệ
    - Không đi ra ngoài ma trận
    - Không đi vào vật cản

- E -> MÔI TRƯỜNG
    - Ma trận 5x5
    - 0 = sạch
    - 1 = dơ
    - 2 = vật cản (không thể di chuyển vào)

- A -> BỘ CHẤP HÀNH
    - Đi lên
    - Đi xuống
    - Đi trái
    - Đi phải
    - Hút bụi

- S -> CẢM BIẾN
    - Kiểm tra ô hiện tại
    - Kiểm tra vị trí hiện tại
    - Kiểm tra biên ma trận
    - Kiểm tra ô kế tiếp có vật cản

## RULE TABLE

IF ô hiện tại dơ
THEN hút bụi

IF ô hiện tại sạch
THEN di chuyển ngẫu nhiên

IF chạm biên trên
THEN bỏ hướng đi lên

IF chạm biên dưới
THEN bỏ hướng đi xuống

IF chạm biên trái
THEN bỏ hướng đi trái

IF chạm biên phải
THEN bỏ hướng đi phải

IF ô kế tiếp có vật cản
THEN bỏ hướng đi đến vật cản đó

```python
import random

# ================= MÔI TRƯỜNG ========================

# 0 = sạch, 1 = dơ, 2 = vật cản
phong = [
    [0,1,0,0,1],
    [0,0,1,2,0],
    [1,0,0,1,0],
    [0,2,0,0,0],
    [1,0,1,0,1]
]

SO_HANG = 5
SO_COT  = 5

# ================= ROBOT =============================

x = 0
y = 0
```

```python
# ================= TẬP LUẬT ==========================

tap_luat = {

    0 : "DI_CHUYEN",
    1 : "HUT_BUI"
}
```

```python
# =====================================================

def hien_thi_phong():

    print("\nTRẠNG THÁI PHÒNG\n")

    for i in range(SO_HANG):

        for j in range(SO_COT):

            if i == x and j == y:
                print("R", end=" ")

            else:
                print(phong[i][j], end=" ")

        print()
```

```python
# =====================================================

def quet_moi_truong():

    print("\nQUÉT MÔI TRƯỜNG")

    print("Vị trí hiện tại :", (x,y))

    print("Giá trị ô hiện tại :", phong[x][y])
```

```python
# =====================================================

def hut_bui():

    print("\nHÀNH ĐỘNG : HÚT BỤI")

    phong[x][y] = 0

    print("Ô hiện tại đã sạch")
```

```python
# =====================================================

def di_chuyen():

    global x, y

    huong = ["w","a","s","d"]

    # ================= KIỂM TRA BIÊN =================

    if x == 0:
        if "w" in huong: huong.remove("w")

    if x == SO_HANG - 1:
        if "s" in huong: huong.remove("s")

    if y == 0:
        if "a" in huong: huong.remove("a")

    if y == SO_COT - 1:
        if "d" in huong: huong.remove("d")

    # ================= KIỂM TRA VẬT CẢN =================

    # Kiểm tra ô phía trên
    if "w" in huong and phong[x-1][y] == 2:
        huong.remove("w")

    # Kiểm tra ô phía dưới
    if "s" in huong and phong[x+1][y] == 2:
        huong.remove("s")

    # Kiểm tra ô phía trái
    if "a" in huong and phong[x][y-1] == 2:
        huong.remove("a")

    # Kiểm tra ô phía phải
    if "d" in huong and phong[x][y+1] == 2:
        huong.remove("d")

    # =================================================

    if not huong:
        print("\nKHÔNG THỂ DI CHUYỂN: Tất cả các hướng đều có vật cản hoặc biên.")
        return # Robot cannot move

    print("\nCÁC HƯỚNG CÓ THỂ ĐI :", huong)

    huong_chon = random.choice(huong)

    print("HƯỚNG ĐƯỢC CHỌN :", huong_chon)

    # ================= ĐI LÊN =================

    if huong_chon == "w":

        x = x - 1

        print("DI CHUYỂN LÊN")

    # ================= ĐI XUỐNG =================

    elif huong_chon == "s":

        x = x + 1

        print("DI CHUYỂN XUỐNG")

    # ================= ĐI TRÁI =================

    elif huong_chon == "a":

        y = y - 1

        print("DI CHUYỂN TRÁI")

    # ================= ĐI PHẢI =================

    elif huong_chon == "d":

        y = y + 1

        print("DI CHUYỂN PHẢI")
```

```python
# =====================================================

def tac_nhan_phan_xa_don_gian():

    # bước 1 : quét môi trường

    quet_moi_truong()

    # bước 2 : lấy trạng thái hiện tại

    trang_thai = phong[x][y]

    # bước 3 : chọn luật

    hanh_dong = tap_luat[trang_thai]

    print("\nLUẬT ĐƯỢC CHỌN :", hanh_dong)

    # ================= HÚT BỤI =================

    if hanh_dong == "HUT_BUI":

        hut_bui()

        print("\nKIỂM TRA LẠI")

        print("Giá trị hiện tại :", phong[x][y])

        di_chuyen()

    # ================= DI CHUYỂN =================

    elif hanh_dong == "DI_CHUYEN":

        di_chuyen()
```

```python
# =====================================================
# CHƯƠNG TRÌNH CHÍNH
# =====================================================

hien_thi_phong()

for buoc in range(15):

    print("\n===================================")
    print("BƯỚC :", buoc + 1)
    print("===================================")

    tac_nhan_phan_xa_don_gian()

    hien_thi_phong()
```
