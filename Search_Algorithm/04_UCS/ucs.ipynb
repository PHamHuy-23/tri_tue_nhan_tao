# ⚖️ Uniform Cost Search (UCS) - 8-Puzzle

File này triển khai giải thuật **Uniform Cost Search (UCS)** (Dijkstra trên đồ thị trạng thái).
UCS mở rộng các nút có chi phí đường đi $g(n)$ nhỏ nhất sử dụng **Priority Queue**.

## 🛠️ 1. Khai báo thư viện & Định nghĩa môi trường

````python
import heapq
import copy
import time

# Trạng thái ban đầu & Mục tiêu
initial_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

def state_to_tuple(state):
    return tuple(tuple(row) for row in state)

def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
    return -1, -1

def generate_children(state):
    children = []
    x, y = find_zero(state)
    moves = [
        ("LÊN", -1, 0, 1),
        ("XUỐNG", 1, 0, 1),
        ("TRÁI", 0, -1, 1),
        ("PHẢI", 0, 1, 1)
    ]
    for move_name, dx, dy, cost in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = copy.deepcopy(state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            children.append((move_name, new_state, cost))
    return children
````

## 🚶‍♂️ 2. Triển khai UCS (Uniform Cost Search)

````python
def ucs(init, goal):
    heap = []
    unique_counter = 0
    
    heapq.heappush(heap, (0, unique_counter, init, []))
    reached = {state_to_tuple(init): 0}
    
    steps = 0
    nodes_expanded = 0
    
    while heap:
        steps += 1
        g_cost, _, current_state, path = heapq.heappop(heap)
        
        if current_state == goal:
            return path, g_cost, steps, nodes_expanded, len(reached)
            
        current_tuple = state_to_tuple(current_state)
        if g_cost > reached.get(current_tuple, float('inf')):
            continue
            
        nodes_expanded += 1
        
        for move, child, step_cost in generate_children(current_state):
            child_tuple = state_to_tuple(child)
            new_g = g_cost + step_cost
            
            if child_tuple not in reached or new_g < reached[child_tuple]:
                reached[child_tuple] = new_g
                unique_counter += 1
                heapq.heappush(heap, (new_g, unique_counter, child, path + [move]))
                
    return None, float('inf'), steps, nodes_expanded, len(reached)
````

## 📊 3. Chạy thử nghiệm thuật toán

````python
print("=== CHẠY UCS (UNIFORM COST SEARCH) ===")
start = time.perf_counter()
path, cost, steps, expanded, reached_count = ucs(initial_state, goal_state)
end = time.perf_counter()
print(f"Đường đi tối ưu: {path}")
print(f"Tổng chi phí đường đi: {cost}")
print(f"Số lần lặp: {steps}")
print(f"Số nút được mở rộng (expanded): {expanded}")
print(f"Số nút được lưu trong Reached Set: {reached_count}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")
````
