# 📉 Iterative Deepening A* (IDA*) Search - 8-Puzzle

IDA* kết hợp tìm kiếm theo chiều sâu lặp lại giới hạn (IDS) và hàm đánh giá $f(n) = g(n) + h(n)$ của A*.
Thay vì lưu toàn bộ các nút trong frontier như A*, IDA* chỉ cần thực hiện DFS với giới hạn ngưỡng $f$ tăng dần, giúp tiết kiệm bộ nhớ đáng kể (chỉ tốn $O(d)$ bộ nhớ thay vì mũ).

## 🛠️ 1. Khai báo và Thiết lập Môi trường

```python
import sys
import time
sys.path.append('../')
from solvers import idastar, h_hamming, h_manhattan, h_euclidean, goal_state

# Cấu hình trạng thái ban đầu
initial_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]
```

## 🔍 2. Chạy thử nghiệm thuật toán IDA*

```python
print("=== IDA* SEARCH (HAMMING) ===")
start = time.perf_counter()
path_ham, limit_ham, steps_ham = idastar(initial_state, goal_state, h_hamming)
end = time.perf_counter()
print(f"Đường đi tối ưu: {path_ham}")
print(f"Ngưỡng f giới hạn cuối: {limit_ham}")
print(f"Tổng số lần gọi nút: {steps_ham}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")

print("=== IDA* SEARCH (MANHATTAN) ===")
start = time.perf_counter()
path_man, limit_man, steps_man = idastar(initial_state, goal_state, h_manhattan)
end = time.perf_counter()
print(f"Đường đi tối ưu: {path_man}")
print(f"Ngưỡng f giới hạn cuối: {limit_man}")
print(f"Tổng số lần gọi nút: {steps_man}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")
```
