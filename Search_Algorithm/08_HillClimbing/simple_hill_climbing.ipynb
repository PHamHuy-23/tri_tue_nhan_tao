# 🧗 Simple Hill Climbing (Leo đồi cơ bản) - 8-Queens

Simple Hill Climbing là thuật toán tìm kiếm cục bộ cơ bản. Tại mỗi bước lặp, thuật toán duyệt ngẫu nhiên các trạng thái lân cận (neighbors) và ngay lập tức dịch chuyển sang trạng thái đầu tiên có giá trị heuristic tốt hơn (ở đây là số lượng cặp tấn công ít hơn).
Thuật toán dừng lại khi không tìm thấy bất kỳ trạng thái lân cận nào tốt hơn trạng thái hiện tại (có thể bị kẹt ở cực trị cục bộ).

## 🛠️ 1. Khai báo Môi trường và Thiết lập

````python
import sys
import random
sys.path.append('./')
from nqueens_solvers import h_attack

# Khởi tạo trạng thái ban đầu ngẫu nhiên cho 8 quân hậu
random.seed(42)
initial_state = [random.randint(0, 7) for _ in range(8)]
print(f"Trạng thái ban đầu: {initial_state}")
print(f"Số cặp quân hậu tấn công ban đầu: {h_attack(initial_state)}")
````

## 🔍 2. Triển khai và Chạy thuật toán Simple Hill Climbing

````python
from nqueens_solvers import simple_hill_climbing

best_state, best_h, steps, history = simple_hill_climbing(initial_state)

print(f"=== KẾT QUẢ SIMPLE HILL CLIMBING ===")
print(f"Trạng thái cuối cùng: {best_state}")
print(f"Số cặp tấn công (Heuristic): {best_h} (mục tiêu là 0)")
print(f"Số bước đi: {steps}")
print(f"Lịch sử Heuristic: {[h for state, h in history]}")
````
