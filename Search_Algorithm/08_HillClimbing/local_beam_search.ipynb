# 📡 Local Beam Search (Tìm kiếm chùm cục bộ) - 8-Queens

Thay vì duy trì một trạng thái đơn lẻ, Local Beam Search theo dõi đồng thời $k$ trạng thái được khởi tạo ngẫu nhiên. Tại mỗi bước lặp, tất cả các lân cận của cả $k$ trạng thái được sinh ra, sau đó thuật toán sắp xếp và chỉ giữ lại $k$ lân cận tốt nhất cho bước tiếp theo.

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
import random
sys.path.append('./')
from nqueens_solvers import h_attack

# Khởi tạo trạng thái ban đầu ngẫu nhiên cho 8 quân hậu
random.seed(42)
initial_state = [random.randint(0, 7) for _ in range(8)]
print(f"Trạng thái ban đầu: {initial_state}")
print(f"Số cặp quân hậu tấn công ban đầu: {h_attack(initial_state)}")
```
