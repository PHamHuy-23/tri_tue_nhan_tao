# 🧗 Steepest Ascent Hill Climbing (Leo đồi dốc nhất) - 8-Queens

Biến thể này cải tiến của leo đồi thông thường bằng cách duyệt **toàn bộ** lân cận và luôn chọn trạng thái có giá trị heuristic tốt nhất (tối ưu nhất) làm trạng thái tiếp theo.
Mặc dù cho chất lượng tốt hơn Simple Hill Climbing, chi phí tính toán mỗi bước sẽ cao hơn và vẫn dễ bị mắc kẹt ở cực trị cục bộ (Local Optimum).

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
import random
sys.path.append('./')
from nqueens_solvers import h_attack

# Khởi tạo trạng thái ban đầu ngẫu nhiên cho 8 quân hậu
random.seed(42)
initial_state = [random.randint(0, 7) for _ in range(8)]
print(f"Trạng thái ban đầu: {initial_state}")
print(f"Số cặp quân hậu tấn công ban đầu: {h_attack(initial_state)}")
```

## 🔍 2. Triển khai và Chạy thuật toán Steepest Ascent Hill Climbing

```python
from nqueens_solvers import steepest_ascent_hill_climbing

best_state, best_h, steps, history = steepest_ascent_hill_climbing(initial_state)

print(f"=== KẾT QUẢ STEEPEST ASCENT HILL CLIMBING ===")
print(f"Trạng thái cuối cùng: {best_state}")
print(f"Số cặp tấn công (Heuristic): {best_h}")
print(f"Số bước đi: {steps}")
print(f"Lịch sử Heuristic: {[h for state, h in history]}")
```
