# 🌡️ Simulated Annealing (Mô phỏng luyện kim) - 8-Queens

Simulated Annealing lấy cảm hứng từ vật lý luyện kim, kết hợp leo đồi và bước đi ngẫu nhiên.
- Nếu trạng thái lân cận tốt hơn, ta **luôn luôn** chấp nhận.
- Nếu lân cận tệ hơn, ta vẫn chấp nhận với xác suất $p = e^{-\frac{\Delta}{T}}$ phụ thuộc vào độ chênh lệch $\Delta$ và nhiệt độ hiện tại $T$ (nhiệt độ giảm dần sau mỗi bước).

## 🛠️ 1. Khai báo Môi trường và Thiết lập

```python
import sys
import random
sys.path.append('./')
from nqueens_solvers import h_attack

# Khởi tạo trạng thái ban đầu ngẫu nhiên cho 8 quân hậu
random.seed(42)
initial_state = [random.randint(0, 7) for _ in range(8)]
print(f"Trạng thái ban đầu: {initial_state}")
print(f"Số cặp quân hậu tấn công ban đầu: {h_attack(initial_state)}")
```

## 🔍 2. Triển khai và Chạy thuật toán Simulated Annealing

```python
from nqueens_solvers import simulated_annealing

best_state, best_h, steps, history = simulated_annealing(initial_state, temp=100.0, alpha=0.95, min_temp=0.01)

print(f"=== KẾT QUẢ SIMULATED ANNEALING ===")
print(f"Trạng thái cuối cùng: {best_state}")
print(f"Số cặp tấn công (Heuristic): {best_h}")
print(f"Số bước đi: {steps}")
print(f"Lịch sử Heuristic: {[h for state, h in history]}")
```
