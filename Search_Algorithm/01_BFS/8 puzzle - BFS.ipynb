## Giải thuật tìm kiếm 8-Puzzle bằng BFS

Đây là triển khai giải thuật Breadth-First Search (BFS) để giải bài toán 8-Puzzle.

### 1. Cài đặt ban đầu

```python
from collections import deque
import copy

# =========================================================
# TRẠNG THÁI BAN ĐẦU
# =========================================================

initial_state = [

    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

# =========================================================
# TRẠNG THÁI MỤC TIÊU
# =========================================================

goal_state = [

    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
```

### 2. Các hàm hỗ trợ

```python
# =========================================================
# IN TRẠNG THÁI PUZZLE
# =========================================================

def print_puzzle(state):

    for row in state:

        print(row)
```

```python
# =========================================================
# CHUYỂN TRẠNG THÁI -> TUPLE
# (để kiểm tra trùng lặp trong set)
# =========================================================

def state_to_tuple(state):

    return tuple(tuple(row) for row in state)
```

```python
# =========================================================
# TÌM Ô TRỐNG (0)
# =========================================================

def find_zero(state):

    for i in range(3):

        for j in range(3):

            if state[i][j] == 0:

                return i, j
```

```python
# =========================================================
# TẠO CÁC TRẠNG THÁI CON
# =========================================================

def generate_children(state):

    children = []

    x, y = find_zero(state)

    moves = [

        ("LÊN", -1, 0),
        ("XUỐNG", 1, 0),
        ("TRÁI", 0, -1),
        ("PHẢI", 0, 1)
    ]

    for move_name, dx, dy in moves:

        nx = x + dx
        ny = y + dy

        # nước đi hợp lệ

        if 0 <= nx < 3 and 0 <= ny < 3:

            new_state = copy.deepcopy(state)

            # hoán đổi

            new_state[x][y], new_state[nx][ny] = \
            new_state[nx][ny], new_state[x][y]

            children.append((move_name, new_state))

    return children
```

```python
# =========================================================
# IN BIÊN (FRONTIER) - (Lưu ý: Hàm này không được sử dụng trong BFS)
# =========================================================

def print_frontier(frontier):

    result = []

    for item in frontier:

        state = item[0]

        result.append(state)

    return result
```

### 3. Thuật toán tìm kiếm BFS

```python
# =========================================================
# TÌM KIẾM BFS
# =========================================================

def bfs():

    # hàng đợi (queue)

    frontier = deque()

    # tập hợp các trạng thái đã duyệt (reached set)

    reached = set()

    # lưu đường đi

    frontier.append((initial_state, []))

    reached.add(state_to_tuple(initial_state))

    step = 1

    # =====================================================

    while frontier:

        print("\n=================================================")
        print("BƯỚC :", step)
        print("=================================================")

        # lấy từ đầu hàng đợi (FIFO)

        current_state, path = frontier.popleft()

        print("\nĐANG DUYỆT NÚT\n")

        print_puzzle(current_state)

        # =================================================
        # KIỂM TRA MỤC TIÊU
        # =================================================

        if current_state == goal_state:

            print("\nĐÃ TÌM THẤY MỤC TIÊU")
            print("\nĐƯỜNG ĐI CUỐI CÙNG :")

            for i, action in enumerate(path, start=1):

                print(f"BƯỚC {i} : {action}")

            print("\nTRẠNG THÁI CUỐI CÙNG\n")

            print_puzzle(current_state)

            return

        # =================================================
        # TẠO CÁC TRẠNG THÁI CON
        # =================================================

        children = generate_children(current_state)

        generated_nodes = []

        for move_name, child_state in children:

            child_tuple = state_to_tuple(child_state)

            # tránh trùng lặp

            if child_tuple not in reached:

                reached.add(child_tuple)

                new_path = path + [move_name]

                frontier.append((child_state, new_path))

                generated_nodes.append((move_name, child_state))

        # =================================================
        # IN CÁC NÚT ĐƯỢC TẠO
        # =================================================

        print("\nBIÊN ĐƯỢC TẠO\n")

        for move_name, node in generated_nodes:

            print("NƯỚC ĐI :", move_name)

            print_puzzle(node)

            print()

        # =================================================
        # IN HÀNG ĐỢI
        # =================================================

        print("\nHÀNG ĐỢI HIỆN TẠI (THỨ TỰ FIFO)\n")

        queue_list = list(frontier)

        for idx, item in enumerate(queue_list):

            print("VỊ TRÍ HÀNG ĐỢI :", idx + 1)

            print_puzzle(item[0])

            print()

        # =================================================
        # IN SỐ LƯỢNG NÚT ĐÃ DUYỆT
        # =================================================

        print("\nSỐ LƯỢNG NÚT ĐÃ DUYỆT :", len(reached))

        step += 1
```

### 4. Chạy thuật toán BFS

```python
# =========================================================
# CHẠY BFS
# =========================================================

bfs()
```
