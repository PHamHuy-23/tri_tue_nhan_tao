# 🥖 Breadth-First Search (BFS) - 8-Puzzle

File này triển khai giải thuật **Breadth-First Search (BFS)** để giải quyết bài toán 8-Puzzle với hai cách tiếp cận:
1. **Cách 1**: Kiểm tra mục tiêu (Goal-test) khi **lấy nút ra** khỏi hàng đợi.
2. **Cách 2**: Kiểm tra mục tiêu (Early Goal Check) ngay khi **sinh nút con**.


## 🛠️ 1. Khai báo thư viện & Định nghĩa môi trường

```python
from collections import deque
import copy
import time

# Trạng thái ban đầu & Mục tiêu
initial_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

def state_to_tuple(state):
    return tuple(tuple(row) for row in state)

def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
    return -1, -1

def generate_children(state):
    children = []
    x, y = find_zero(state)
    moves = [
        ("LÊN", -1, 0),
        ("XUỐNG", 1, 0),
        ("TRÁI", 0, -1),
        ("PHẢI", 0, 1)
    ]
    for move_name, dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = copy.deepcopy(state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            children.append((move_name, new_state))
    return children
```

## 🚶‍♂️ 2. Triển khai BFS - Cách 1 (Goal-test on POP)

```python
def bfs_way_1(init, goal):
    frontier = deque([(init, [])])
    reached = {state_to_tuple(init)}
    steps = 0
    nodes_expanded = 0
    
    while frontier:
        steps += 1
        current_state, path = frontier.popleft()
        nodes_expanded += 1
        
        # Kiểm tra mục tiêu khi lấy ra khỏi hàng đợi
        if current_state == goal:
            return path, steps, nodes_expanded, len(reached)
            
        for move, child in generate_children(current_state):
            child_tuple = state_to_tuple(child)
            if child_tuple not in reached:
                reached.add(child_tuple)
                frontier.append((child, path + [move]))
                
    return None, steps, nodes_expanded, len(reached)
```

## ⚡ 3. Triển khai BFS - Cách 2 (Early Goal Check)

```python
def bfs_way_2(init, goal):
    # Kiểm tra mục tiêu ngay trạng thái ban đầu
    if init == goal:
        return [], 0, 0, 1
        
    frontier = deque([(init, [])])
    reached = {state_to_tuple(init)}
    steps = 0
    nodes_expanded = 0
    
    while frontier:
        steps += 1
        current_state, path = frontier.popleft()
        nodes_expanded += 1
        
        for move, child in generate_children(current_state):
            # Kiểm tra mục tiêu khi sinh nút con
            if child == goal:
                reached.add(state_to_tuple(child))
                return path + [move], steps, nodes_expanded, len(reached)
                
            child_tuple = state_to_tuple(child)
            if child_tuple not in reached:
                reached.add(child_tuple)
                frontier.append((child, path + [move]))
                
    return None, steps, nodes_expanded, len(reached)
```

## 📊 4. Chạy thử nghiệm và So sánh kết quả

```python
print("=== CHẠY BFS CÁCH 1 (GOAL-TEST ON POP) ===")
start = time.perf_counter()
path1, steps1, expanded1, reached1 = bfs_way_1(initial_state, goal_state)
end = time.perf_counter()
print(f"Đường đi: {path1}")
print(f"Số bước lặp: {steps1}")
print(f"Số nút được mở rộng (expanded): {expanded1}")
print(f"Số nút được lưu trong Reached Set: {reached1}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")

print("=== CHẠY BFS CÁCH 2 (EARLY GOAL CHECK) ===")
start = time.perf_counter()
path2, steps2, expanded2, reached2 = bfs_way_2(initial_state, goal_state)
end = time.perf_counter()
print(f"Đường đi: {path2}")
print(f"Số bước lặp: {steps2}")
print(f"Số nút được mở rộng (expanded): {expanded2}")
print(f"Số nút được lưu trong Reached Set: {reached2}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")
```
