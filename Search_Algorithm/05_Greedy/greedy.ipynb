# 🧭 Greedy Best-First Search - 8-Puzzle

Greedy Best-First Search là giải thuật tìm kiếm có thông tin (Informed Search).
Nó lựa chọn mở rộng nút $n$ dựa trên hàm mục tiêu:
$$f(n) = h(n)$$
Trong đó $h(n)$ là giá trị heuristic ước lượng chi phí từ trạng thái hiện tại đến đích.

Trong notebook này, ta sẽ giải bài toán 8-Puzzle bằng Greedy Best-First Search sử dụng 3 hàm heuristic:
1. **Hamming Distance** (Số ô đặt sai vị trí)
2. **Manhattan Distance** (Tổng khoảng cách đi theo trục ngang/dọc)
3. **Euclidean Distance** (Tổng khoảng cách đường chim bay)


## 🛠️ 1. Khai báo và Thiết lập Môi trường

```python
import sys
import time
sys.path.append('../')
from solvers import greedy, h_hamming, h_manhattan, h_euclidean, goal_state

# Cấu hình trạng thái ban đầu
initial_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]
```

## 🔍 2. Chạy thử nghiệm với các Heuristics khác nhau

```python
print("=== GREEDY BEST-FIRST SEARCH (HAMMING) ===")
start = time.perf_counter()
path_ham, steps_ham, exp_ham, reach_ham = greedy(initial_state, goal_state, h_hamming)
end = time.perf_counter()
print(f"Đường đi: {path_ham}")
print(f"Số bước lặp: {steps_ham}, Số nút mở rộng: {exp_ham}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")

print("=== GREEDY BEST-FIRST SEARCH (MANHATTAN) ===")
start = time.perf_counter()
path_man, steps_man, exp_man, reach_man = greedy(initial_state, goal_state, h_manhattan)
end = time.perf_counter()
print(f"Đường đi: {path_man}")
print(f"Số bước lặp: {steps_man}, Số nút mở rộng: {exp_man}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")

print("=== GREEDY BEST-FIRST SEARCH (EUCLIDEAN) ===")
start = time.perf_counter()
path_euc, steps_euc, exp_euc, reach_euc = greedy(initial_state, goal_state, h_euclidean)
end = time.perf_counter()
print(f"Đường đi: {path_euc}")
print(f"Số bước lặp: {steps_euc}, Số nút mở rộng: {exp_euc}")
print(f"Thời gian chạy: {(end - start)*1000:.4f} ms\n")
```
