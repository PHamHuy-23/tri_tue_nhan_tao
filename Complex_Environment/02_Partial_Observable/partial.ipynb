## 🛠️ 1. Khai báo và Cập nhật Belief State

```python
import sys
sys.path.append('../../Search_Algorithm')
from complex_solvers import update_belief_partial, ALL_STATES

# Khởi đầu: Tin rằng có thể ở bất kỳ trạng thái nào trong 8 trạng thái
current_belief = frozenset(ALL_STATES)
print(f"Số lượng trạng thái khả dĩ ban đầu: {len(current_belief)}")

# Hành động 1: Robot đi sang TRÁI và cảm biến nhận thấy ô Trái Bẩn -> Percept: (0, 1)
current_belief = update_belief_partial(current_belief, "TRÁI", (0, 1))
print(f"Sau khi rẽ TRÁI và thấy BẨN, số trạng thái khả dĩ còn lại: {len(current_belief)}")
for s in current_belief:
    print(s)

# Hành động 2: HÚT bụi, sau đó cảm biến thấy ô Trái Sạch -> Percept: (0, 0)
current_belief = update_belief_partial(current_belief, "HÚT", (0, 0))
print(f"\nSau khi HÚT và thấy SẠCH, số trạng thái khả dĩ còn lại: {len(current_belief)}")
for s in current_belief:
    print(s)
```
