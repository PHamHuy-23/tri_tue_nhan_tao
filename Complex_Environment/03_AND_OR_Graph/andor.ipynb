# 🌲 AND-OR Graph Search (Môi trường không xác định)

Trong môi trường không xác định, kết quả hành động của tác nhân là ngẫu nhiên/không chắc chắn (ví dụ: hành động HÚT có thể làm sạch phòng, hoặc đôi khi bị kẹt không làm sạch được).
Tác nhân sử dụng thuật toán **AND-OR Graph Search** để xây dựng một **Kế hoạch điều kiện (Conditional Plan)**: xử lý tất cả các kết quả khả dĩ có thể xảy ra của hành động.

## 🛠️ 1. Thiết lập và Chạy AND-OR Graph Search

```python
import sys
sys.path.append('../../Search_Algorithm')
from complex_solvers import and_or_graph_search

# Trạng thái ban đầu: Robot ở bên Trái (0), cả 2 phòng đều Bẩn (1, 1)
initial_state = (0, 1, 1)
plan = and_or_graph_search(initial_state)

def print_plan(plan, indent=0):
    if plan == []:
        print(" -> Đạt mục tiêu!")
        return
    if plan is None:
        print(" -> Thất bại (Không có kế hoạch)")
        return
    
    action, branches = plan
    print(f"Thực hiện [{action}]")
    for state, subplan in branches.items():
        print("  " * (indent + 1) + f"Nếu kết quả là {state}: ", end="")
        print_plan(subplan, indent + 1)

print("=== KẾT QUẢ KẾ HOẠCH ĐIỀU KIỆN (CONDITIONAL PLAN) ===")
print_plan(plan)
```
